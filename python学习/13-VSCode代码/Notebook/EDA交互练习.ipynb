```python
from pathlib import Path
import pandas as pd
root = Path.cwd().parents[1] if Path.cwd().name == 'Notebook' else Path.cwd()
df = pd.read_csv(root / '12-示例数据' / '环境监测数据.csv', parse_dates=['日期'])
df.head()
```

```python
df.info()
df.isna().mean().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
df.groupby('日期')['PM2.5'].mean().plot(figsize=(9, 4), title='Daily PM2.5')
plt.show()
```
